```python
from pyspark.sql import SparkSession
from pyspark.sql.types import StructType, StructField, IntegerType, StringType
# Initialize Spark session
spark = SparkSession.builder.appName("UserDataProcessing").getOrCreate()
# Step 1: Load u.data into new RDD
u_data_rdd = spark.sparkContext.textFile("/content/ratings.csv")
# Step 2: Convert RDD to DataFrame
u_data_schema = StructType([
    StructField("userId", IntegerType(), True),
    StructField("item_id", IntegerType(), True),
    StructField("rating", IntegerType(), True),
    StructField("timestamp", IntegerType(), True)
])
u_data_df = u_data_rdd.map(lambda x: x.split("\t")).map(lambda x: (int(x[0]), int(x[1]), int(x[2]), int(x[3]))).toDF(schema=u_data_schema)
# Step 3: Return schema
u_data_df.printSchema()

```

```python
from pyspark.sql import SparkSession

# Initialize Spark session
spark = SparkSession.builder.appName("MovieRatings").getOrCreate()

# Step 1: Load CSV file with correct options
u_data_df = spark.read.csv("/content/ratings.csv", header=True, inferSchema=True)

# Step 2: Register DataFrame as a temp table
u_data_df.createOrReplaceTempView("u_data")

# Step 3: Display contents of u_data table
spark.sql("SELECT * FROM u_data").show()

# Step 4: Show number of items reviewed by each user
spark.sql("SELECT userId, COUNT(movieId) AS num_reviews FROM u_data GROUP BY userId ORDER BY num_reviews DESC").show()

# Step 5: Show number of users who reviewed each item
spark.sql("SELECT movieId, COUNT(userId) AS num_users FROM u_data GROUP BY movieId ORDER BY num_users DESC").show()
```
